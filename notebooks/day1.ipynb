```sql
%python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("THAMES").getOrCreate()

print("Spark is ready!")
```

```sql
%python
data = [
    ("Rahul", "Shoes", 3000),
    ("Tojo", "Shirt", 1500),
    ("Rahul", "Watch", 2000),
    ("Trish", "Laptop", 60000),
    ("Tojo", "Shoes", 4000)
]

columns = ["Customer", "Product", "Amount"]

df = spark.createDataFrame(data, columns)

display(df)
```

```sql
%python
df.filter(df.Amount > 2000).show()
from pyspark.sql.functions import sum

df.groupBy("Customer").agg(
    sum("Amount").alias("Total_Spending")
).show()
```

```sql
%python
data = [
    ("Rahul", "Shoes", 3000, "Bangalore"),
    ("Rush", "Shirt", 1500, "Delhi"),
    ("Rahul", "Watch", 2000, "Bangalore"),
    ("Priya", "Laptop", 60000, "Mumbai"),
    ("Rush", "Shoes", 4000, "Delhi")
]

columns = ["Customer", "Product", "Amount", "City"]

df = spark.createDataFrame(data, columns)

display(df)
```

```sql
%python
df.select("Customer", "Amount").show()
df2 = df.withColumn("Amount_With_Tax", df.Amount * 1.18)

display(df2)

df.filter(df.Amount > 3000).show()

df.orderBy(df.Amount.desc()).show()
```

```sql
%python
print("Total transactions:", df.count())
df.show(3)
```

```sql
%python
filtered_df = df.filter(df.Amount > 3000)

selected_df = filtered_df.select("Customer", "Amount")
```

```sql
%python
from pyspark.sql.functions import sum

customer_spending = (
    df.groupBy("Customer")
      .agg(sum("Amount").alias("Total_Spending"))
      .filter("Total_Spending > 5000")
)

display(customer_spending)
```
