```python
from pyspark.sql.functions import sum, when, col
from pyspark.ml.feature import VectorAssembler, StandardScaler

# 1. Product catalogue
products = [
    ("P001", "Linen Casual Shirt", "Fashion", "Shirts", "Brand A", 1499, "Sky Blue", "Linen", "Relaxed", 4.3),
    ("P002", "Oxford Formal Shirt", "Fashion", "Shirts", "Brand B", 1999, "White", "Cotton", "Slim", 4.1),
    ("P003", "Running Sneakers", "Footwear", "Shoes", "Brand C", 3499, "Black", "Mesh", "Regular", 4.5),
    ("P004", "Everyday Hoodie", "Fashion", "Hoodies", "Brand A", 1799, "Grey", "Cotton", "Oversized", 4.2),
    ("P005", "Wireless Headphones", "Electronics", "Audio", "Brand D", 5999, "Black", "Plastic", "Over-ear", 4.4),
    ("P006", "Smartwatch", "Electronics", "Wearables", "Brand E", 8999, "Silver", "Metal", "Regular", 4.0)
]

product_columns = [
    "Product_ID", "Product_Name", "Category",
    "Subcategory", "Brand", "Price", "Colour",
    "Material", "Fit", "Rating"
]

product_df = spark.createDataFrame(products, product_columns)

# 2. Customer activity
customer_data = [
    ("C101", "P001", "View"),
    ("C101", "P001", "Cart"),
    ("C101", "P002", "View"),
    ("C101", "P004", "Purchase"),
    ("C101", "P003", "View"),
    ("C102", "P005", "View"),
    ("C102", "P005", "Cart"),
    ("C102", "P006", "Purchase"),
    ("C102", "P005", "Purchase"),
    ("C103", "P003", "View"),
    ("C103", "P003", "Purchase"),
    ("C103", "P001", "View"),
    ("C104", "P002", "View"),
    ("C104", "P002", "Cart"),
    ("C104", "P001", "Purchase"),
    ("C105", "P006", "View"),
    ("C105", "P005", "View"),
    ("C105", "P006", "Cart")
]

activity_columns = ["Customer_ID", "Product_ID", "Action"]

activity_df = spark.createDataFrame(customer_data, activity_columns)

# 3. Join customer activity with product catalogue
customer_products = activity_df.join(
    product_df,
    on="Product_ID",
    how="inner"
)

# 4. Customer behaviour features
customer_features = customer_products.groupBy(
    "Customer_ID"
).agg(
    sum(when(col("Action") == "View", 1).otherwise(0)).alias("Total_Views"),
    sum(when(col("Action") == "Cart", 1).otherwise(0)).alias("Total_Carts"),
    sum(when(col("Action") == "Purchase", 1).otherwise(0)).alias("Total_Purchases")
)

# 5. Customer spending
purchase_features = customer_products.filter(
    col("Action") == "Purchase"
).groupBy("Customer_ID").agg(
    sum("Price").alias("Total_Spending")
)

# 6. Combine features
final_features = customer_features.join(
    purchase_features,
    on="Customer_ID",
    how="left"
).fillna(0)

# 7. Assemble ML features
assembler = VectorAssembler(
    inputCols=[
        "Total_Views",
        "Total_Carts",
        "Total_Purchases",
        "Total_Spending"
    ],
    outputCol="raw_features"
)

assembled_df = assembler.transform(final_features)

# 8. Scale features
scaler = StandardScaler(
    inputCol="raw_features",
    outputCol="features",
    withStd=True,
    withMean=False
)

scaler_model = scaler.fit(assembled_df)

scaled_df = scaler_model.transform(assembled_df)

# Final verification
print("Products:", product_df.count())
print("Customer activities:", activity_df.count())
print("Customer records:", final_features.count())

display(scaled_df)
```

```python
from pyspark.sql.functions import col, row_number, desc
from pyspark.sql.window import Window

# 1. Identify previously purchased products
purchased_products = (
    activity_df
    .filter(col("Action") == "Purchase")
    .select("Customer_ID", "Product_ID")
    .distinct()
)

# 2. Identify each customer's preferred category
preference_window = Window.partitionBy("Customer_ID").orderBy(
    desc("Category_Score")
)

top_categories = (
    category_preferences
    .withColumn("rank", row_number().over(preference_window))
    .filter(col("rank") == 1)
    .select("Customer_ID", "Category", "Category_Score")
)

# 3. Match preferred categories with products
candidate_products = top_categories.join(
    product_df,
    on="Category",
    how="inner"
)

# 4. Exclude previously purchased products
recommendation_candidates = candidate_products.join(
    purchased_products,
    on=["Customer_ID", "Product_ID"],
    how="left_anti"
)

# 5. Rank products based on ratings
recommendation_window = Window.partitionBy("Customer_ID").orderBy(
    desc("Rating")
)

recommendations = (
    recommendation_candidates
    .withColumn(
        "Recommendation_Rank",
        row_number().over(recommendation_window)
    )
    .filter(col("Recommendation_Rank") <= 3)
)

# 6. Display final recommendations
display(
    recommendations.select(
        "Customer_ID",
        "Product_ID",
        "Product_Name",
        "Category",
        "Price",
        "Rating",
        "Recommendation_Rank"
    ).orderBy("Customer_ID", "Recommendation_Rank")
)
```
